## Q1. Program to Insert and Delete an Element at a Given Position in a List


```python
# Insert and Delete element at given position
lst = [10, 20, 30, 40, 50]
print("Original List:", lst)

# Insert
pos = int(input("Enter position to insert (0-based index): "))
element = int(input("Enter element to insert: "))
if pos >= 0 and pos <= len(lst):
    lst.insert(pos, element)
    print("List after insertion:", lst)
else:
    print("Invalid position for insertion")
    
# Delete
pos = int(input("Enter position to delete (0-based index): "))
if pos >= 0 and pos < len(lst):
    deleted_element = lst.pop(pos)
    print("Deleted element:", deleted_element)
    print("List after deletion:", lst)
    print("List after deletion:", lst)
else:
    print("Invalid position for deletion")
    

```

## Q2. Program to Search an Element in a List


```python
# Search element in a list
lst = [5, 15, 25, 35, 45]
print("List:", lst)
key = int(input("Enter element to search: "))
if key in lst:
     print("Element found at index:", lst.index(key))
else:
     print("Element not found")







```

## Q3.  Program to Demonstrate Basic Operations on Data Structures (List)


```python
# Basic operations on list
lst = []
# Insertion
lst.append(10)
lst.append(20)
lst.append(30)
print("After insertion:", lst)
# Traversal
print("Traversing list:")
for item in lst:
    print(item, end=" ")
# Searching
key = 20
if key in lst:
    print("\nElement", key, "found")
else:
    print("\nElement not found")
# Deletion
lst.remove(20)
print("After deletion:", lst)

# Length
print("Length of list:", len(lst))





```

## Q4.  Program to Create and Traverse One-Dimensional Array


```python
# One-dimensional array
arr = []
n = int(input("Enter number of elements: "))
for i in range(n):
    element = int(input("Enter element: "))
    arr.append(element)
    print("Array elements are:")
    for i in range(len(arr)):
        print("Index", i, "=", arr[i])



```

## Q5. Program to Represent Two-Dimensional Array Using Python List


```python
# Two-dimensional array
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))
matrix = []
for i in range(rows):
    row = []
    for j in range(cols):
        element = int(input(f"Enter element at position ({i},{j}): "))
        row.append(element)
    matrix.append(row)
print("2D Array (Matrix):")
for i in range(rows):
    for j in range(cols):
            print(matrix[i][j], end=" ")
    print()



```
